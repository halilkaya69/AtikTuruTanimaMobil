```python
from ultralytics import YOLO

model = YOLO('yolov12n.pt')

print(f"YOLOv12- Nano Modeli Yüklendi. Eğitim Başlıyor...")

# 2. Eğitimi Başlat
results = model.train(
    data='data.yaml',
    epochs=50,
    imgsz=512,
    batch=16,         # Bellek yönetimi için düşük tutuyoruz
    workers=4,
    device=0,        # GPU
    #patience=30,     # Erken durdurma
    #augment=True,    # Veri artırımı
    name='atik_yolov12_nano',
    exist_ok=True
)

print("\nEğitim Tamamlandı! Sonuçlar: runs/detect/atik_yolov12_nano/")
```

```python
from ultralytics import YOLO

model = YOLO("runs/detect/atik_yolov12_nano/weights/best.pt")

metrics = model.val(
    data="data.yaml",
    split="test"   # <-- TEST SET
)

print(metrics)
```
